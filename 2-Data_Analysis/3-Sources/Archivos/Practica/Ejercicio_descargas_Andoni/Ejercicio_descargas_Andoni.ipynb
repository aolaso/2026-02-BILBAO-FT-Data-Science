## Ejercicio descargas

Python es un lenguaje de propósito general, por lo que podemos desarrollar otros programas,
aparte de ciencia de datos, que nos permitan montar dashboards, crear APIs o automatizar
procesos. En este caso vas a automatizar un proceso, en el que tendrás que desarrollar un
script que ordene todos los archivos de una carpeta, dependiendo de la extensión de los
mismos.


El objetivo es ordenar la típica carpeta de cajón desastre donde se nos acumulan archivos de
todos los formatos como: imágenes, words, presentaciones, excels, ejecutables, zips, etc... Y
que no vamos a ir viendo archivo a archivo, ordenándolos.... La solución suele ser eliminarlo
todo. Vamos a intentar no tener que llegar a eso, ordenando los ficheros automáticamente en
carpetas y revisar las que nos interesen. La típica carpeta donde aplicaríamos este programa es
la de "Descargas", donde se suelen acumular muchos archivos temporales.


Por tanto, **el programa tiene que guardar los archivos de la carpeta "Descargas" (o cualquier
otra donde acumules mucho archivo temporal) en los siguientes directorios dentro de la
propia carpeta "Descargas"**:
- Imagenes
- Documentos
- Software
- Otros

Cada vez que ejecutes el script, automáticamente se ordenarán todos los archivos en sus
correspondientes carpetas.


````python
doc_types = ('.doc', '.docx', '.txt', '.pdf', '.xls', '.ppt', '.xlsx', '.pptx')
img_types = ('.jpg', '.jpeg', '.png', '.svg', '.gif')
software_types = ('.exe', '.py','.ipynb')
````

### 1. Crea un un programa con el flujo principal y encapsúlalo en funciones.

````python
import os
import shutil
````

````python
# Las extensiones que nos dan en el enunciado para clasificar cada archivo
doc_types = ('.doc', '.docx', '.txt', '.pdf', '.xls', '.ppt', '.xlsx', '.pptx')
img_types = ('.jpg', '.jpeg', '.png', '.svg', '.gif')
software_types = ('.exe', '.py', '.ipynb')

# Los nombres de las subcarpetas donde moveremos los archivos
CARPETAS = ["Documentos", "Imagenes", "Software", "Otros"]


def crear_carpetas(ruta_base):
    """
    Creamos las subcarpetas (Documentos, Imagenes, Software, Otros)
    dentro de la carpeta que le pasemos.
    exist_ok=True hace que si la carpeta ya existe, no dé error.
    """
    for carpeta in CARPETAS:
        # os.path.join() une rutas de forma segura
        ruta_carpeta = os.path.join(ruta_base, carpeta)
        # os.makedirs() crea la carpeta
        os.makedirs(ruta_carpeta, exist_ok=True)


def clasificar_archivo(nombre_archivo):
    """
    Dado el nombre de un archivo, devolvemos a qué carpeta pertenece
    según su extensión.
    """
    # os.path.splitext() separa nombre y extensión: "informe.pdf" -> ("informe", ".pdf")
    _, extension = os.path.splitext(nombre_archivo)
    extension = extension.lower()
    
    if extension in doc_types:
        return "Documentos"
    elif extension in img_types:
        return "Imagenes"
    elif extension in software_types:
        return "Software"
    else:
        return "Otros"


def ordenar_carpeta(ruta_base):
    """
    Función principal: recorre todos los archivos de la carpeta,
    los clasifica y los mueve a su subcarpeta correspondiente.
    No borra nada, solo mueve.
    """
    # Primero creamos las subcarpetas
    crear_carpetas(ruta_base)
    
    # Recorremos todo lo que hay en la carpeta
    for elemento in os.listdir(ruta_base):
        ruta_completa = os.path.join(ruta_base, elemento)
        
        # Solo movemos ARCHIVOS, no carpetas
        if os.path.isfile(ruta_completa):
            carpeta_destino = clasificar_archivo(elemento)
            destino = os.path.join(ruta_base, carpeta_destino, elemento)
            # shutil.move() mueve el archivo (como arrastrar y soltar en Finder)
            shutil.move(ruta_completa, destino)
            print(f"  {elemento}  ->  {carpeta_destino}/")
    
    print("\nOrdenación completada")
````

````python
# Ejecutamos sobre nuestra carpeta de Descargas real
# No se borra nada: solo se mueven archivos sueltos a subcarpetas
# Las carpetas que ya existan dentro de Descargas no se tocan
ruta_descargas = "/Users/andoni/Downloads"
ordenar_carpeta(ruta_descargas)
````
