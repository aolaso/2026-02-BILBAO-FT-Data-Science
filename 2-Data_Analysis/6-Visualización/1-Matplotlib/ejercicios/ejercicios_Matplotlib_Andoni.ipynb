2. Sabemos que podemos pintar gráficas de dos formas: la figura incluye los ejes o teniendo figura y ejes por separado.

Usando solo una figura, usa numpy para los valores del eje X entre 0 y 5. 

Pinta dos gráficas en dos cajas distintas, a la izquierda una recta con pendiente positiva de 3 que pase por (0,0) y a la derecha una recta con pendiente negativa de 3 que pase por (0,-5). 

Elige la precisión en el eje X que desees.  

```python
# Creamos 10 puntos equidistantes entre 0 y 5 para el eje X
# linspace(inicio, fin, cuántos_puntos) nos da una lista de números espaciados uniformemente
x = np.linspace(0, 5, 10)
```

```python
# RECTA 1: pendiente +3, pasa por (0, 0)
# La ecuación de una recta es: y = m*x + b
# m = 3 (pendiente)
# Pasa por (0, 0), es decir: cuando x=0, y=0
# Sustituimos: 0 = 3*0 + b → b = 0
# Nos queda: y = 3*x
y1 = 3 * x

# RECTA 2: pendiente -3, pasa por (0, -5)
# m = -3 (pendiente negativa, la recta baja)
# Pasa por (0, -5), es decir: cuando x=0, y=-5
# Sustituimos: -5 = -3*0 + b → b = -5
# Nos queda: y = -3*x - 5
y2 = -3 * x - 5
```

```python
# Ahora las dibujamos en dos cajas distintas (subplots) dentro de la misma figura
# plt.subplot(filas, columnas, posición) crea una cuadrícula de gráficos
# (1, 2, 1) = 1 fila, 2 columnas, y ahora dibujamos en la posición 1 (izquierda)
plt.subplot(1, 2, 1)
# Dibujamos la recta que sube
plt.plot(x, y1)

# Cambiamos a la posición 2 (derecha) de la misma cuadrícula
plt.subplot(1, 2, 2)
# Dibujamos la recta que baja
plt.plot(x, y2)
```

3. Fija el eje X entre 0 y 5 y el eje Y entre -15 y 15

```python
# Dibujamos el gráfico de la izquierda (posición 1)
plt.subplot(1, 2, 1)
plt.plot(x, y1)
# Fijamos que el eje X solo muestre de 0 a 5
plt.xlim(0, 5)
# Fijamos que el eje Y solo muestre de -15 a 15
# Así ambos gráficos tienen la misma escala y podemos compararlos visualmente
plt.ylim(-15, 15)

# Dibujamos el gráfico de la derecha (posición 2)
plt.subplot(1, 2, 2)
plt.plot(x, y2)
# Ponemos los mismos límites para que ambos gráficos se vean iguales
plt.xlim(0, 5)
plt.ylim(-15, 15);
```

4. Llama al eje X "eje X", al eje Y "eje Y" y pon de títulos "recta sube" y "recta baja". Muestra dos etiquetas de ejeX pero solo una de eje Y.

```python
# --- GRÁFICO IZQUIERDO ---
plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.xlim(0, 5)
plt.ylim(-15, 15)
# Ponemos una etiqueta al eje X para que quien mire el gráfico sepa qué representa
plt.xlabel("Eje x")
# Ponemos etiqueta al eje Y (solo en el gráfico izquierdo, como pide el enunciado)
plt.ylabel("Eje y")
# Ponemos un título arriba del gráfico
plt.title("Recta sube")

# --- GRÁFICO DERECHO ---
plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.xlim(0, 5)
plt.ylim(-15, 15)
# Etiqueta del eje X (el enunciado pide dos etiquetas de eje X, una por gráfico)
plt.xlabel("Eje x")
# NO ponemos ylabel porque el enunciado dice "solo una de eje Y" (la del izquierdo)
plt.title("Recta baja");
```

5. Usando una figura Y EJES POR SEPARADO, usa numpy para los valores del eje X entre 0 y 5. Pinta dos gráficas en dos cajas distintas, a la izquierda una recta con pendiente positiva de 3 que pase por (0,0) y a la derecha una recta con pendiente negativa de 3 que pase por (0,-5). Elige la precisión en el eje X que desees.  

```python
# plt.subplots(1, 2) crea una figura con 2 gráficos lado a lado
# Nos devuelve DOS cosas:
#   fig = la figura entera (el lienzo)
#   ax = una lista con los 2 ejes (ax[0] es el izquierdo, ax[1] es el derecho)
# Esta es la interfaz "orientada a objetos": tenemos los ejes como variables
# y les decimos directamente qué hacer, en vez de depender de "cuál está activo"
fig, ax = plt.subplots(1, 2)

# --- GRÁFICO IZQUIERDO (ax[0]) ---
# Dibujamos la recta que sube en el eje izquierdo
ax[0].plot(x, y1)
# En esta interfaz, los métodos llevan "set_" delante: set_xlim, set_xlabel, set_title...
ax[0].set_xlim(0, 5)
ax[0].set_ylim(-15, 15)
ax[0].set_xlabel("Eje x")
ax[0].set_ylabel("Eje y")
ax[0].set_title("Recta sube")

# --- GRÁFICO DERECHO (ax[1]) ---
# Dibujamos la recta que baja en el eje derecho
ax[1].plot(x, y2)
ax[1].set_xlim(0, 5)
ax[1].set_ylim(-15, 15)
ax[1].set_xlabel("Eje x")
# No ponemos set_ylabel porque el enunciado solo quiere una etiqueta Y
ax[1].set_title("Recta baja");
```

7. Crea un gráfico de barras. Pintamos la altura de un grupo de amigos, cada barra representa a una persona. Ana mide 160 cm, Luis mide 180 cm, Pedro mide 175 cm, Sofía mide 190 cm, Carmen mide 170 cm. Las barras serán verdes.
Consejo: si no vas a tener que manipular especialmente los ejes, es más sencillo dejarlos dentro de la figura.

```python
# Creamos la lista de nombres (serán las etiquetas del eje X)
personas = ["Ana", "Luis", "Pedro", "Sofía", "Carmen"]
# Creamos la lista de alturas (serán la altura de cada barra)
alturas = [160, 180, 175, 190, 170]

# Dibujamos las barras: en el eje X los nombres, en el Y las alturas
# color="green" hace que todas las barras sean verdes
plt.bar(personas, alturas, color="green")
# Etiquetamos los ejes y ponemos un título
plt.xlabel("Personas")
plt.ylabel("Alturas")
plt.title("Comparación alturas grupo")
# Limitamos el eje Y entre 140 y 200 para que se vea mejor la diferencia
# Si empezáramos desde 0, todas las barras se verían casi iguales
# porque 160 y 190 son "casi lo mismo" comparado con 0
plt.ylim(140, 200);
```

```python
# BONUS: ordenamos las barras de mayor a menor para que se vea más claro
# Creamos una Serie de pandas con las alturas como valores y los nombres como índice
# sort_values(ascending=False) las ordena de mayor a menor
serie = pd.Series(alturas, index=personas).sort_values(ascending=False)

# Ahora dibujamos usando la serie ordenada
# serie.index = los nombres ordenados (Sofía, Luis, Pedro, Carmen, Ana)
# serie.values = las alturas ordenadas (190, 180, 175, 170, 160)
plt.bar(serie.index, serie.values, color="green")
plt.xlabel("Personas")
plt.ylabel("Alturas")
plt.title("Comparación alturas grupo")
plt.ylim(140, 200);
```

8. Cambia el tamaño de la figura a (3,5)

```python
# plt.figure(figsize=(ancho, alto)) crea una figura con un tamaño personalizado
# Los números son pulgadas: (3, 5) = 3 pulgadas de ancho, 5 de alto
# Una figura más alta que ancha para un gráfico de barras verticales
plt.figure(figsize=(3, 5))
plt.bar(serie.index, serie.values, color="green")
plt.xlabel("Personas")
plt.ylabel("Alturas")
plt.title("Comparación alturas grupo")
plt.ylim(140, 200);
```

10. Prueba a hacer el gráfico con las barras en horizontal.

```python
# Cambiamos el figsize: ahora más ancho que alto (5, 3) porque las barras van en horizontal
plt.figure(figsize=(5, 3))
# plt.barh() es igual que plt.bar() pero con las barras en HORIZONTAL
# "barh" = "bar horizontal"
# Los nombres van en el eje Y, las alturas en el eje X
plt.barh(serie.index, serie.values, color="green")
# Como las barras son horizontales, los ejes se intercambian:
# ahora "Alturas" va en X y "Personas" en Y
plt.xlabel("Alturas")
plt.ylabel("Personas")
plt.title("Comparación alturas grupo")
# Limitamos el eje X (que ahora es donde están los valores numéricos)
plt.xlim(140, 200)
plt.xticks([160, 190])
plt.yticks(["Sofía", "Ana"], rotation=-45, fontsize=8);
```

11. Prueba a invertir los ejes con ax.invert_axis()
Necesitarás tener los ejes disponibles fuera de la figura

```python
# Aquí necesitamos la interfaz OO (fig, ax) porque vamos a usar ax.invert_xaxis()
# que es un método que solo existe en los objetos Axes, no en plt directamente
fig, ax = plt.subplots()
# Dibujamos las barras horizontales
plt.barh(serie.index, serie.values)
# Invertimos el eje X: ahora los valores altos están a la izquierda y los bajos a la derecha
ax.invert_xaxis()
# Invertimos el eje Y: ahora el primer elemento está abajo y el último arriba
ax.invert_yaxis()
```

13. Vamos a introducir pandas. Crea un DataFrame con las columnas "Year" de valores 2015, 2016, 2017, 2018, 2019 y la columna "Sold_items_A" de valores 1000, 3500, 4000, 5500, 7000

```python
# Creamos un DataFrame con dos columnas:
# "year" = los años, como strings (texto), no como números
# "sold_items_a" = las unidades vendidas por el departamento A cada año
# Pasamos un diccionario donde cada clave es el nombre de la columna
# y cada valor es la lista de datos de esa columna
df_13 = pd.DataFrame({
    "year": ["2015", "2016", "2017", "2018", "2019"],
    "sold_items_a": [1000, 3500, 4000, 5500, 7000]
})

# Mostramos el DataFrame para ver que se ha creado bien
df_13
```

14. Ahora pintamos un gráfico de línea con las ventas respecto al año. Pon un título y etiquetas en los ejes.
Cambia el estilo a 'seaborn-white'. La línea debe ser con rayas y verde.

```python
# Cambiamos el estilo a uno con fondo blanco limpio (para dejar atrás el dark_background)
plt.style.use("seaborn-v0_8-white")

# Dibujamos una línea: eje X = años, eje Y = ventas
# '--g' es un atajo que combina estilo de línea + color:
#   '--' = línea discontinua (rayas)
#   'g' = green (verde)
plt.plot(df_13['year'], df_13['sold_items_a'], '--g')
# Ponemos título y etiquetas para que el gráfico sea legible
plt.title("Evolución ventas anual")
plt.xlabel("Años")
plt.ylabel("Ventas del departamento A (unidades)");
```

15. Otro departamento B ha vendido en esos años 2000, 3100, 5000, 4000, 6000 unidades. Incluye esa columna en el DataFrame y pinta en la misma gráfica las dos líneas. B es una línea punteada y roja. Muestra una leyenda abajo a la derecha.

```python
# Añadimos una nueva columna al DataFrame con las ventas del departamento B
df_13['sold_items_b'] = [2000, 3100, 5000, 4000, 6000]
```

```python
# Dibujamos las dos líneas en el MISMO gráfico
# (al llamar a plt.plot() dos veces seguidas, se dibujan en la misma figura)

# Línea del departamento A: discontinua y verde
# label="Departamento A" → le ponemos un nombre para la leyenda
plt.plot(df_13['year'], df_13['sold_items_a'], '--g', label="Departamento A")

# Línea del departamento B: punteada y roja
# ':r' = punteada (:) + roja (r)
plt.plot(df_13['year'], df_13['sold_items_b'], ':r', label="Departamento B")

plt.title("Evolución ventas anual")
plt.xlabel("Años")
plt.ylabel("Ventas departamentales (unidades)")
# plt.legend() dibuja la cajita de leyenda
# Recoge automáticamente los "label" que pusimos en cada plot
# loc="lower right" la coloca abajo a la derecha
plt.legend(loc="lower right");
```

19. Vamos a pintar un histograma de unas alturas generadas aleatoriamente. Crea 300 alturas entre 150 y 190 cm.

Pista: https://pandas.pydata.org/pandas-docs/version/0.23.4/generated/pandas.DataFrame.plot.html

```python
# np.random.randint(150, 190, 300) genera 300 números enteros aleatorios entre 150 y 189
# (el 190 NO se incluye, es el límite superior exclusivo)
# Simulamos las alturas de 300 personas
alturas_random = np.random.randint(150, 190, 300)

# Lo convertimos en una Serie de pandas para poder usar el método .plot()
serie_19 = pd.Series(alturas_random)

# Dibujamos un histograma
# kind="hist" → tipo histograma (barras que cuentan cuántos valores caen en cada rango)
# bins=10 → dividimos el rango en 10 intervalos/barras
# Cada barra muestra cuántas personas tienen una altura dentro de ese rango
serie_19.plot(kind="hist", bins=10)
```
